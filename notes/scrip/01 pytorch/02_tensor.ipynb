## 一、创建

| 概念     | NumPy        | PyTorch                         | 区别                 |
| -------- | ------------ | ------------------------------- | -------------------- |
| 多维数组 | `np.ndarray` | `torch.Tensor`                  | 多一个 `device` 属性 |
| 数据类型 | `np.float32` | `torch.float32`                 | 名称不同，语义相同   |
| 求导     | 无           | `requires_grad`                 | PyTorch 独有         |
| 内存共享 | `np.asarray` | `torch.from_numpy` / `.numpy()` | 双向共享内存         |

```python
import torch 
a = torch.tensor([[1,2,3],[45,65,85]]) 
a
```

```python
b = torch.tensor([[2,3,6],[96,78,96]],dtype=torch.float32)
b
```

```python
c = torch.zeros(2,3)
c
```

```python
d = torch.ones(3,4)
d
```

```python
e = torch.randn(2,5)   # 标准正态
e
```

```python
f = torch.rand(1,4)         # [0,1) 均匀分布
f
```

```python
h = torch.empty(2,3)         # 不初始化（快，但内容随机，别直接读）
h
```

```python
# 3) 从 NumPy 来（共享内存，改一个另一个也变）
import numpy as np
arr = np.arange(6, dtype="float32").reshape(2, 3)
t = torch.from_numpy(arr)
t
```

## 二、运算

| 写法           | 语义                 | 用途                 |
| -------------- | -------------------- | -------------------- |
| `A * B`        | 逐元素乘（Hadamard） | 门控、掩码           |
| `A @ B`        | 矩阵乘               | 线性层               |
| `torch.einsum` | 任意缩并             | 注意力、复杂张量网络 |
| `A.sum() * B`  | 标量乘               | —                    |


 

```python
a = torch.tensor([1,2,3])
b = torch.tensor([[2],[2],[2]])
c = torch.tensor([2,2,2])
a
```
